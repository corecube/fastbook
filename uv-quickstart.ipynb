# Using `uv` for this project

This project is managed with [uv](https://docs.astral.sh/uv/). Dependencies live in `pyproject.toml`, the exact resolved versions in `uv.lock`, and the virtual environment in `.venv/`.

## Setup (once)

Recreate the environment from the lockfile:

```bash
uv sync
```

This creates `.venv/` and installs everything pinned in `uv.lock`. No manual `pip install` needed.

## Running things

Run any command inside the project env without activating it:

```bash
uv run python script.py
uv run jupyter lab        # launch notebooks
```

Or activate the venv the classic way:

```bash
source .venv/bin/activate
```

## Managing dependencies

```bash
uv add fastbook          # add a package (updates pyproject.toml + uv.lock + installs)
uv remove fastbook       # remove a package
uv pip list              # list installed packages
```

Always use `uv add` / `uv remove` rather than `pip install` so `pyproject.toml` and `uv.lock` stay in sync.

## Jupyter kernel

Run notebooks against this project's env by selecting the `python3` kernel at:

```
.venv/share/jupyter/kernels/python3
```

If you change dependencies, **restart the kernel** so it picks up the new packages.

## Note for this repo

The project was renamed to `fastbook-study` in `pyproject.toml` because uv won't let a project depend on its own name (`fastbook` is also the PyPI package the notebooks import).
